# Vortex Panel Method in 2D

**Name:** *(Saiyam Mishra)*  **Roll no.:** *(SC26M016)*  **Course:** *(Aerodynamics)*

In this report I use the vortex panel method from Anderson, *Fundamentals of Aerodynamics*, Section 4.10, to find the flow around a cylinder and around NACA airfoils. I wrote two versions:

* **Method 0:** each panel has a constant vortex strength $\gamma_j$.
* **Method 1:** the vortex strength changes linearly along each panel.

Symbols (same as Anderson):
* $(X_j, Y_j)$ are the boundary points (panel corners). They go clockwise, starting at the trailing edge and going along the lower surface first.
* $(x_i, y_i)$ are the control points (panel midpoints).
* $S_j$ is the panel length and $\phi_j$ is the panel angle.
* $\beta_i = \phi_i + \pi/2 - \alpha$ is the angle between the freestream and the outward normal.
* $J_{ij}$ and $K_{ij}$ are the integrals that tell how much panel $j$ affects control point $i$.

## 1. Algorithms

For a point on panel $j$, $x_j = X_j + s\cos\phi_j$ and $y_j = Y_j + s\sin\phi_j$, with $s$ going from 0 to $S_j$. The integrals are

$$J_{ij} = \int_0^{S_j} \frac{(x_i - x_j)\cos\phi_i + (y_i - y_j)\sin\phi_i}{(x_i-x_j)^2 + (y_i-y_j)^2}\,ds, \qquad
K_{ij} = \int_0^{S_j} \frac{(x_i - x_j)\sin\phi_i - (y_i - y_j)\cos\phi_i}{(x_i-x_j)^2 + (y_i-y_j)^2}\,ds$$

### Algorithm for Method 0 (constant strength)

1. From the boundary points, find $S_j$, $\phi_j$, the control points and $\beta_j$ for every panel.
2. For every $i \ne j$, calculate $J_{ij}$ and $K_{ij}$ with numerical integration. For $i = j$ they are 0.
3. Write the no-flow-through condition at every control point:
   $\sum_j J_{ij}\gamma_j = 2\pi V_\infty \cos\beta_i$.
4. Replace the last equation with the Kutta condition $\gamma_1 + \gamma_N = 0$.
5. Solve the equations for $\gamma_j$.
6. Surface velocity: $V_i = V_\infty\sin\beta_i - \sum_j \frac{\gamma_j}{2\pi}K_{ij} + \frac{\gamma_i}{2}$, and $C_{p,i} = 1 - (V_i/V_\infty)^2$.
7. Lift: $\Gamma = \sum_j \gamma_j S_j$, $c_l = 2\Gamma / (V_\infty c)$.

For the cylinder there is no sharp trailing edge, so instead of the Kutta condition I add the condition $\Gamma = \sum_j \gamma_j S_j = 0$ and solve with least squares.

### Algorithm for Method 1 (linear strength)

1. Same geometry as Method 0.
2. On panel $j$ the strength is $\gamma(s) = \gamma_j(1 - s/S_j) + \gamma_{j+1}\,s/S_j$. Now the unknowns are the $N+1$ values at the boundary points.
3. For every $i \ne j$, calculate $J_{ij}$ and also $J'_{ij}$, which is the same integral multiplied by $s/S_j$. Panel $j$ adds $J_{ij} - J'_{ij}$ to the coefficient of $\gamma_j$ and $J'_{ij}$ to the coefficient of $\gamma_{j+1}$.
4. For $i = j$ (a panel acting on its own midpoint), add $+1$ to the coefficient of $\gamma_i$ and $-1$ to the coefficient of $\gamma_{i+1}$.
5. Write the $N$ no-flow-through equations, and add the Kutta condition $\gamma_1 + \gamma_{N+1} = 0$ as one extra equation. Now there are $N+1$ equations for $N+1$ unknowns.
6. Solve for the $\gamma$ values.
7. Surface velocity at each boundary point is $V = \gamma$, so $C_p = 1 - (\gamma/V_\infty)^2$.
8. Lift: $\Gamma = \sum_j S_j(\gamma_j + \gamma_{j+1})/2$, $c_l = 2\Gamma/(V_\infty c)$.

## 2. Python code

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
```

### Geometry

`panels` finds the panel lengths, angles and control points. `cylinder` and `naca4` make the boundary points. For the airfoil, `cosine` spacing puts more points near the nose and the tail, and `uniform` spacing puts them at equal steps in $x$.

```python
def panels(XB, YB):
    N = len(XB) - 1
    S = np.zeros(N)
    phi = np.zeros(N)
    XC = np.zeros(N)
    YC = np.zeros(N)
    for j in range(N):
        dx = XB[j+1] - XB[j]
        dy = YB[j+1] - YB[j]
        S[j] = np.sqrt(dx**2 + dy**2)
        phi[j] = np.arctan2(dy, dx)
        XC[j] = (XB[j] + XB[j+1]) / 2
        YC[j] = (YB[j] + YB[j+1]) / 2
    return XC, YC, S, phi


def cylinder(N):
    angle = np.linspace(0, -2*np.pi, N+1)
    return np.cos(angle), np.sin(angle)


def naca4(digits, N, spacing='cosine'):
    m = int(digits[0]) / 100
    p = int(digits[1]) / 10
    t = int(digits[2:]) / 100
    n = N // 2
    if spacing == 'cosine':
        x = 0.5 * (1 - np.cos(np.linspace(0, np.pi, n+1)))
    else:
        x = np.linspace(0, 1, n+1)
    yt = 5*t*(0.2969*np.sqrt(x) - 0.1260*x - 0.3516*x**2 + 0.2843*x**3 - 0.1036*x**4)
    yc = np.zeros(n+1)
    slope = np.zeros(n+1)
    if m > 0:
        for k in range(n+1):
            if x[k] < p:
                yc[k] = m/p**2 * (2*p*x[k] - x[k]**2)
                slope[k] = 2*m/p**2 * (p - x[k])
            else:
                yc[k] = m/(1-p)**2 * ((1 - 2*p) + 2*p*x[k] - x[k]**2)
                slope[k] = 2*m/(1-p)**2 * (p - x[k])
    th = np.arctan(slope)
    xu = x - yt*np.sin(th)
    yu = yc + yt*np.cos(th)
    xl = x + yt*np.sin(th)
    yl = yc - yt*np.cos(th)
    XB = np.concatenate([xl[::-1], xu[1:]])
    YB = np.concatenate([yl[::-1], yu[1:]])
    return XB, YB
```

### Method 0: constant strength

`JK` calculates the integrals $J_{ij}$ and $K_{ij}$ with `quad`.

**About the `fix` option.** With the plain solve (`fix=False`), I found that $C_p$ on the airfoils jumps up and down from panel to panel (shown in Section 4). This happens because the pattern $\gamma = +a, -a, +a, \dots$ gives almost no flow through the surface, so the matrix is almost singular and this pattern gets amplified. With `fix=True`, the system is solved with an SVD and the one pattern with the smallest singular value is left out. This idea came from the AI assistant (see Section 7).

```python
def JK(XC, YC, XB, YB, S, phi):
    N = len(XC)
    J = np.zeros((N, N))
    K = np.zeros((N, N))
    for i in range(N):
        for j in range(N):
            if i == j:
                continue
            def fJ(s):
                dx = XC[i] - (XB[j] + s*np.cos(phi[j]))
                dy = YC[i] - (YB[j] + s*np.sin(phi[j]))
                return (dx*np.cos(phi[i]) + dy*np.sin(phi[i])) / (dx**2 + dy**2)
            def fK(s):
                dx = XC[i] - (XB[j] + s*np.cos(phi[j]))
                dy = YC[i] - (YB[j] + s*np.sin(phi[j]))
                return (dx*np.sin(phi[i]) - dy*np.cos(phi[i])) / (dx**2 + dy**2)
            J[i, j] = quad(fJ, 0, S[j])[0]
            K[i, j] = quad(fK, 0, S[j])[0]
    return J, K


def vortex_panel_method_0(XB, YB, alpha_deg, Vinf=1.0, kutta=True, fix=True):
    """Constant-strength vortex panel method.
    Input:  boundary points XB, YB, angle of attack in degrees, Vinf,
            kutta=True for airfoil / False for cylinder, fix=True removes the sawtooth.
    Output: dictionary with control points, gamma, V, Cp, Gamma and Cl."""
    alpha = np.radians(alpha_deg)
    XC, YC, S, phi = panels(XB, YB)
    beta = phi + np.pi/2 - alpha
    N = len(XC)
    J, K = JK(XC, YC, XB, YB, S, phi)
    b = 2*np.pi*Vinf*np.cos(beta)

    if kutta:
        A = J.copy()
        A[N-1, :] = 0
        A[N-1, 0] = 1
        A[N-1, N-1] = 1
        b[N-1] = 0
        if fix:
            U, sig, Vt = np.linalg.svd(A)
            gamma = np.zeros(N)
            for k in range(N-1):
                gamma = gamma + (U[:, k] @ b) / sig[k] * Vt[k, :]
        else:
            gamma = np.linalg.solve(A, b)
    else:
        A = np.vstack([J, S])
        b = np.append(b, 0)
        gamma = np.linalg.lstsq(A, b, rcond=None)[0]

    V = np.zeros(N)
    for i in range(N):
        V[i] = Vinf*np.sin(beta[i]) - np.sum(gamma*K[i, :])/(2*np.pi) + gamma[i]/2
    Cp = 1 - (V/Vinf)**2

    Gamma = np.sum(gamma*S)
    chord = max(XB) - min(XB)
    Cl = 2*Gamma/(Vinf*chord)
    return {'XC': XC, 'YC': YC, 'gamma': gamma, 'V': V, 'Cp': Cp, 'Gamma': Gamma, 'Cl': Cl}
```

### Method 1: linear strength

This uses the same integrand for $J$, but each panel now fills two columns of the matrix.

```python
def vortex_panel_method_1(XB, YB, alpha_deg, Vinf=1.0, kutta=True):
    """Linear-strength vortex panel method.
    Input:  boundary points XB, YB, angle of attack in degrees, Vinf,
            kutta=True for airfoil / False for cylinder.
    Output: dictionary with boundary points, gamma at the boundary points, Cp, Gamma and Cl."""
    alpha = np.radians(alpha_deg)
    XC, YC, S, phi = panels(XB, YB)
    beta = phi + np.pi/2 - alpha
    N = len(XC)

    A = np.zeros((N+1, N+1))
    b = np.zeros(N+1)
    for i in range(N):
        b[i] = 2*np.pi*Vinf*np.cos(beta[i])
        for j in range(N):
            if i == j:
                A[i, j] = A[i, j] + 1
                A[i, j+1] = A[i, j+1] - 1
                continue
            def f(s):
                dx = XC[i] - (XB[j] + s*np.cos(phi[j]))
                dy = YC[i] - (YB[j] + s*np.sin(phi[j]))
                return (dx*np.cos(phi[i]) + dy*np.sin(phi[i])) / (dx**2 + dy**2)
            Jij = quad(f, 0, S[j])[0]
            Jpij = quad(lambda s: s/S[j]*f(s), 0, S[j])[0]
            A[i, j] = A[i, j] + Jij - Jpij
            A[i, j+1] = A[i, j+1] + Jpij

    if kutta:
        A[N, 0] = 1
        A[N, N] = 1
    else:
        for j in range(N):
            A[N, j] = A[N, j] + S[j]/2
            A[N, j+1] = A[N, j+1] + S[j]/2

    gamma = np.linalg.solve(A, b)
    Cp = 1 - (gamma/Vinf)**2

    Gamma = 0
    for j in range(N):
        Gamma = Gamma + S[j]*(gamma[j] + gamma[j+1])/2
    chord = max(XB) - min(XB)
    Cl = 2*Gamma/(Vinf*chord)
    return {'XB': XB, 'YB': YB, 'gamma': gamma, 'Cp': Cp, 'Gamma': Gamma, 'Cl': Cl}
```

### Checking the panels

The red arrows are the normals. They should all point out of the body. If they point inward, the points are in the wrong order.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
XB, YB = cylinder(16)
XC, YC, S, phi = panels(XB, YB)
ax[0].plot(XB, YB, 'ko-')
ax[0].quiver(XC, YC, -np.sin(phi), np.cos(phi), color='r')
ax[0].set_title('Cylinder, 16 panels')
ax[0].axis('equal')
XB, YB = naca4('4412', 30)
XC, YC, S, phi = panels(XB, YB)
ax[1].plot(XB, YB, 'ko-', ms=3)
ax[1].quiver(XC, YC, -np.sin(phi), np.cos(phi), color='r')
ax[1].set_title('NACA 4412, 30 panels')
ax[1].axis('equal')
plt.show()
```

## 3. Verification on a circular cylinder

The exact answer is $C_p = 1 - 4\sin^2\theta$ and the exact vortex strength is $\gamma = 2V_\infty\sin\theta$.

```python
XB, YB = cylinder(16)
r0 = vortex_panel_method_0(XB, YB, 0, kutta=False)
r1 = vortex_panel_method_1(XB, YB, 0, kutta=False)
th0 = np.arctan2(r0['YC'], r0['XC'])
th1 = np.arctan2(YB, XB)
th = np.linspace(-np.pi, np.pi, 200)

plt.plot(np.degrees(th), 1 - 4*np.sin(th)**2, 'k-', label='exact')
plt.plot(np.degrees(th0), r0['Cp'], 'ro', label='method 0')
plt.plot(np.degrees(th1), r1['Cp'], 'bs', mfc='none', label='method 1')
plt.xlabel('theta (deg)')
plt.ylabel('Cp')
plt.title('Cylinder, 16 panels')
plt.legend()
plt.grid()
plt.show()
print('Gamma method 0 =', round(r0['Gamma'], 10), '   Gamma method 1 =', round(r1['Gamma'], 10))
```

### Convergence

I compare $\gamma$ with the exact $2\sin\theta$ and double the number of panels each time.

```python
Ns = [16, 32, 64, 128]
err0 = []
err1 = []
for N in Ns:
    XB, YB = cylinder(N)
    r0 = vortex_panel_method_0(XB, YB, 0, kutta=False)
    th0 = np.arctan2(r0['YC'], r0['XC'])
    err0.append(max(abs(r0['gamma'] - 2*np.sin(th0))))
    r1 = vortex_panel_method_1(XB, YB, 0, kutta=False)
    th1 = np.arctan2(YB, XB)
    err1.append(max(abs(r1['gamma'] - 2*np.sin(th1))))
    print('N =', N, '  error method 0 =', '%.2e' % err0[-1], '  error method 1 =', '%.2e' % err1[-1])

slope0 = np.polyfit(np.log(Ns), np.log(err0), 1)[0]
slope1 = np.polyfit(np.log(Ns), np.log(err1), 1)[0]
plt.loglog(Ns, err0, 'o-', label='method 0, slope = %.2f' % slope0)
plt.loglog(Ns, err1, 's-', label='method 1, slope = %.2f' % slope1)
plt.xlabel('number of panels N')
plt.ylabel('max error in gamma')
plt.legend()
plt.grid(which='both')
plt.show()
```

**Discussion.** Both methods give the exact $C_p$ curve and zero circulation. For Method 0 the $C_p$ at the control points is exact even with 16 panels, which is a lucky effect of the regular polygon, so I used $\gamma$ to check convergence instead. When I double $N$, the Method 0 error halves (slope about $-1$, first order), and the Method 1 error becomes about 8 times smaller (slope about $-2.9$). Method 1 with 16 panels is already better than Method 0 with 128 panels.

One problem I had: for the cylinder with an even number of panels, replacing one equation with $\Gamma = 0$ gave a "singular matrix" error, because the sawtooth pattern $(+a, -a, \dots)$ also gives zero normal velocity. Adding $\Gamma = 0$ as an extra row and using least squares solved this.

## 4. Verification on airfoils

### The jumping $C_p$ in Method 0

```python
XB, YB = naca4('0015', 100)
plain = vortex_panel_method_0(XB, YB, 4, fix=False)
fixed = vortex_panel_method_0(XB, YB, 4, fix=True)
lin = vortex_panel_method_1(XB, YB, 4)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(plain['XC'], plain['Cp'], 'r.-', lw=0.5)
ax[0].set_title('Method 0, plain solve')
ax[1].plot(fixed['XC'], fixed['Cp'], 'b.', label='method 0 with fix')
ax[1].plot(lin['XB'], lin['Cp'], 'k-', label='method 1')
ax[1].set_title('Method 0 with fix, and method 1')
ax[1].legend()
for a in ax:
    a.invert_yaxis()
    a.set_xlabel('x/c')
    a.set_ylabel('Cp')
    a.grid()
plt.show()
print('Cl plain =', round(plain['Cl'], 4), '  Cl fixed =', round(fixed['Cl'], 4), '  Cl method 1 =', round(lin['Cl'], 4))
```

### Number of panels and panel spacing

Angle of attack 4°. As the reference I use Method 1 with 320 cosine-spaced panels. This cell takes about a minute.

```python
Ns = [20, 40, 80, 160]
fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
for k, digits in enumerate(['0015', '4412']):
    XB, YB = naca4(digits, 320)
    Cl_ref = vortex_panel_method_1(XB, YB, 4)['Cl']
    print('NACA', digits, '  reference Cl =', round(Cl_ref, 4))
    for spacing in ['cosine', 'uniform']:
        for name in ['method 0', 'method 1']:
            Cls = []
            for N in Ns:
                XB, YB = naca4(digits, N, spacing)
                if name == 'method 0':
                    Cls.append(vortex_panel_method_0(XB, YB, 4)['Cl'])
                else:
                    Cls.append(vortex_panel_method_1(XB, YB, 4)['Cl'])
            print('   ', spacing, name, '  Cl =', np.round(Cls, 4))
            err = abs(np.array(Cls) - Cl_ref)
            style = '-' if spacing == 'cosine' else '--'
            mark = 'o' if name == 'method 0' else 's'
            ax[k].loglog(Ns, err, style + mark, label=name + ', ' + spacing)
    ax[k].set_title('NACA ' + digits)
    ax[k].set_xlabel('N')
    ax[k].set_ylabel('|Cl - Cl_ref|')
    ax[k].legend(fontsize=8)
    ax[k].grid(which='both')
plt.show()
```

**Discussion.**
* **Method 1 with cosine spacing works best.** Its $c_l$ is within about 1.3 % of the reference even with 20 panels, and the error drops about 4 times each time $N$ doubles (second order).
* **Method 0 needs many panels.** With 20 panels it gives $c_l = 0.63$ for the 0015 instead of 0.49, and 1.21 for the 4412 instead of 1.00. From about 80 panels on it is within about 0.5 % of the reference.
* **The plain solve of Method 0 gives a jumping $C_p$**, down to about $-410$ for the 0015 at 4°, even though $c_l$ is fine (0.4957 against 0.4937 from Method 1). $c_l$ is a sum over all panels, so the up-and-down errors cancel, but $C_p$ shows them. After the fix, Method 0 agrees with Method 1.
* **Uniform spacing is bad.** Near the nose the airfoil is very curved, and equal steps in $x$ give a few long panels there. Method 1 then converges more slowly, and Method 0 does not converge at all ($c_l$ wanders between 0.55 and 0.84 for the 0015, and between 0.79 and 1.27 for the 4412). That is why I use cosine spacing for the rest of the report.

## 5. Airfoil characteristics

NACA 0015 (symmetric) and NACA 4412 (cambered), 100 panels, cosine spacing.

```python
alphas = [-2, 0, 2, 4, 6, 8, 10]
fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
for k, digits in enumerate(['0015', '4412']):
    XB, YB = naca4(digits, 100)
    Cl0 = []
    Cl1 = []
    for a in alphas:
        Cl0.append(vortex_panel_method_0(XB, YB, a)['Cl'])
        Cl1.append(vortex_panel_method_1(XB, YB, a)['Cl'])
    fit = np.polyfit(np.radians(alphas), Cl1, 1)
    a0 = fit[0]
    alpha_L0 = -np.degrees(fit[1]/fit[0])
    print('NACA', digits, ':  lift slope =', round(a0, 3), 'per rad  =', round(a0*np.pi/180, 4),
          'per deg,   zero-lift angle =', round(alpha_L0, 2), 'deg')
    aL0_thin = 0 if digits == '0015' else -4.15
    ax[k].plot(alphas, 2*np.pi*np.radians(np.array(alphas) - aL0_thin), 'k--', label='thin airfoil theory')
    ax[k].plot(alphas, Cl0, 'o', mfc='none', label='method 0')
    ax[k].plot(alphas, Cl1, 's', mfc='none', label='method 1')
    ax[k].set_title('NACA ' + digits)
    ax[k].set_xlabel('alpha (deg)')
    ax[k].set_ylabel('Cl')
    ax[k].legend()
    ax[k].grid()
plt.show()
```

### Comparison with experiment
Experimental values from Abbott & von Doenhoff, *Theory of Wing Sections*, Appendix IV ($Re = 6\times10^6$). *(Fill in the experimental column from the book.)*

| | lift slope, panel method (per deg) | lift slope, experiment (per deg) | $\alpha_{L0}$, panel method | $\alpha_{L0}$, experiment | $c_{l,\max}$, experiment |
|---|---|---|---|---|---|
| NACA 0015 | 0.123 | | 0.00° | | |
| NACA 4412 | 0.120 | | −4.33° | | |

**Discussion.**
* The lift slope is 7.05 per radian (0.123 per degree) for the 0015 and 6.86 per radian (0.120 per degree) for the 4412. Both are more than thin airfoil theory ($2\pi = 6.28$) because the airfoils have thickness. The formula $2\pi(1 + 0.77\,t/c)$ gives 7.01 for 15 % and 6.86 for 12 % thickness, which agrees well.
* For the 0015 the zero-lift angle is 0°, as it should be for a symmetric airfoil. For the 4412 it is −4.33°, close to the thin airfoil theory value of −4.15°. The 4412 has twice the camber of a 2412, so its zero-lift angle is about twice as large.
* $c_l$ is a straight line for all angles. The panel method has no viscosity, so it cannot predict stall or drag. Real airfoils stall at around 14–16°, and their lift slope is a bit lower than the predicted one.
* In the $C_p$ plots, the suction peak near the nose grows quickly as $\alpha$ increases (at 10° it reaches about $-4.7$ on the 0015 and $-5.5$ on the 4412). The 4412 already has a large gap between the upper and lower surfaces at $\alpha = 0°$ ($c_l \approx 0.52$), which is why it makes lift at zero angle. At the trailing edge the upper and lower $C_p$ meet, so the Kutta condition is working.

## 6. Limitations: effect of thickness

I ran Method 1 for NACA 0006 up to 0030 (symmetric airfoils with 6 % to 30 % thickness).

```python
thick = [6, 9, 12, 15, 18, 21, 24, 30]
a0_list = []
for t in thick:
    XB, YB = naca4('00%02d' % t, 100)
    Cl_at_0 = vortex_panel_method_1(XB, YB, 0)['Cl']
    Cl_at_8 = vortex_panel_method_1(XB, YB, 8)['Cl']
    a0 = (Cl_at_8 - Cl_at_0) / np.radians(8)
    a0_list.append(a0)
    print('NACA 00%02d   lift slope = %.3f per rad' % (t, a0))

tc = np.array(thick) / 100
plt.plot(tc, a0_list, 's-', label='panel method (method 1)')
plt.plot(tc, 2*np.pi*(1 + 0.77*tc), 'k:', label='2 pi (1 + 0.77 t/c)')
plt.axhline(2*np.pi, color='k', ls='--', label='thin airfoil theory')
plt.xlabel('t/c')
plt.ylabel('lift slope (per rad)')
plt.legend()
plt.grid()
plt.show()
```

```python
h = 50
rise = []
for t in thick:
    XB, YB = naca4('00%02d' % t, 100)
    r = vortex_panel_method_1(XB, YB, 6)
    xu = XB[h:]
    cpu = r['Cp'][h:]
    rise.append(np.interp(0.9, xu, cpu) - np.interp(0.3, xu, cpu))
    if t in [6, 12, 18, 24, 30]:
        plt.plot(xu, cpu, label='NACA 00%02d' % t)
plt.gca().invert_yaxis()
plt.ylim(1, -6)
plt.xlabel('x/c')
plt.ylabel('Cp (upper surface)')
plt.title('alpha = 6 deg')
plt.legend()
plt.grid()
plt.show()

for t, d in zip(thick, rise):
    print('NACA 00%02d   pressure rise from 0.3c to 0.9c on upper surface = %.2f' % (t, d))
```

**Discussion.**
* The panel method says the lift slope **goes up** with thickness, from 6.6 per radian at 6 % to 7.9 per radian at 30 %. In real experiments (Abbott & von Doenhoff) the lift slope **goes down** for thick airfoils. So the error of the panel method gets bigger as the airfoil gets thicker.
* The reason is the boundary layer. At $\alpha = 6°$, the pressure rise that the upper-surface flow has to climb between $0.3c$ and $0.9c$ grows from 0.50 (6 %) to 1.60 (30 %). This adverse pressure gradient slows the boundary layer down, so on thick airfoils it gets thick and separates near the trailing edge. Then the flow no longer leaves the trailing edge smoothly, which is exactly what the Kutta condition assumes.
* Thin airfoils have a very sharp suction peak at the nose instead ($C_p \approx -6.8$ for the 0006 at 6°, against about $-2.2$ for the 0030), so they stall from the leading edge. The panel method cannot predict either kind of stall, and it always gives zero drag.
* **Conclusion:** the panel method works well for thin airfoils at small angles of attack. For thick airfoils, viscous effects become important and a boundary-layer method (like the one in XFOIL) is needed.
